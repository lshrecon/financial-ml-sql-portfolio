```
import pymysql
```

```
import os

conn = pymysql.connect(
    host=os.environ['MYSQL_HOST'],
    user=os.environ['MYSQL_USER'],
    password=os.environ['MYSQL_PASSWORD'],
    database=os.environ.get('MYSQL_DATABASE', 'stock_data'),
    charset='utf8mb4',
)
```

```
# The following cell previews a separately prepared financial_data table.
```

```
# Preview only 10 rows; the original unbounded fetch printed the entire table.
try:
    with conn.cursor() as cur:
        cur.execute('SELECT * FROM financial_data LIMIT %s', (10,))
        for record in cur.fetchall():
            print(record)
finally:
    conn.close()
```
